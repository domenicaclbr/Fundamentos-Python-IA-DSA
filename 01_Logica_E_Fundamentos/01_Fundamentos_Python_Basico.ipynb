#### Exercício 2

Atribuíção de valor a variáveis e uso de operadores aritméticos.

```python
# Solução

numero1 = 100
numero2 = 200

print(f"{numero1} + {numero2} é igual a {numero1 + numero2}")
print(f"{numero1} - {numero2} é igual a {numero1 - numero2}")
print(f"{numero1} * {numero2} é igual a {numero1 * numero2}")
print(f"{numero1} / {numero2} é igual a {numero1 / numero2:.2f}")
```

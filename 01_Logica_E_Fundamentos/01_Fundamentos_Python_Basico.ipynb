#### Exercício 3 

Objetivo: Diferenciar uma variável de escopo local e uma de escopo global 

Resposta: A Variável de escopo local pode ser usada somente dentro de sua própria função, diferente de uma de escopo global que pode ser usada livremente dentro do código.

```python
# Exemplo prático de escopo 

var_global = "Eu sou global"

def funcao_teste():
    var_local = "Eu sou local"
    print(var_local)

funcao_teste()
print(var_global)
```

#### Exercício 30

Objetivo: Integração de entrada de dados (input), conversão de tipos (type casting), processamento aritmético e formatação de saída (string formatting).

```python
# Receber altura e peso do usuário (em float), calcular o IMC (Índice de Massa Corporal) usando a fórmula IMC = peso / (altura * altura) e imprimir o resultado com duas casas decimais
# Solução 

altura = float(input("Digite a sua altura (ex.: 1.68): "))
peso = float(input("Digite o seu peso (ex.: 68.5): "))
imc = peso / (altura * altura)

print(f"Seu Índice de Massa Corporal (IMC) é: {imc:.2f}")
```

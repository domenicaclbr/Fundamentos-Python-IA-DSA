#### Exercício 2

Objetivo: Atribuíção de valor a variáveis e uso de operadores aritméticos.

```python
# Solução

numero1 = 100
numero2 = 200

print(f"{numero1} + {numero2} é igual a {numero1 + numero2}")
print(f"{numero1} - {numero2} é igual a {numero1 - numero2}")
print(f"{numero1} * {numero2} é igual a {numero1 * numero2}")
print(f"{numero1} / {numero2} é igual a {numero1 / numero2:.2f}")
```

#### Exercício 3 

Objetivo: Diferenciar uma variável de escopo local e uma de escopo global 

Resposta: A Variável de escopo local pode ser usada somente dentro de sua própria função, diferente de uma de escopo global que pode ser usada livremente dentro do código.

```python
# Exemplo prático de escopo 

var_global = "Eu sou global"

def funcao_teste():
    var_local = "Eu sou local"
    print(var_local)

funcao_teste()
print(var_global)
```

#### Exercício 7 

Objetivo: Utilização do operador de módulo (%) para identificação de paridade em números inteiros através de lógica condicional.

```python
# Solução

numero = int(input("Digite um número:"))

if numero % 2 == 0:
    print(f"O número {numero} é PAR!")
else: 
    print(f"O número {numero} é ÍMPAR!")
```

### Estudo de caso 2 - Pedra, Papel ou Tesoura

Neste estudo de caso, o intuito é criar o passo a passo do jogo "Pedra, Papel e Tesoura", desde a escolha de cada item até o resultado final, com a vitória de um dos jogadores ou empate. Este estudo de caso serve para colocar em prática alguns dos ensinamentos sobre o fundamento da linguagem Python e, também, lógica de programação. 


#### Criação do Pseudocódigo para o Jogo

Antes do código, é recomendado criar um pseudocódigo, a linha de raciocínio do código, mais conhecido como "Lógica de Programação". O professor deu as seguintes instruções para a construção do pseudocódigo:

- Pode ser ultilizado: variáveis, entrada e saída

Regras de Vitória:
- 🪨 Pedra vence a ✂️ tesoura (quebrando-a)
- 🧻 Papel vence a 🪨 Pedra (cobrindo-a)
- ✂️ Tesoura vence o 🧻 papel (cortando-o)

Se ambos jogadores escolherem a mesma coisa, haverá empate.

##### Pseudocódigo ⬇️

// 1 Introdução

ESCREVA "---- Jogo Pedra, Papel e Tesoura (2 jogadores)"

ESCREVA "Regras: Escolham entre "pedra", "papel" ou "tesoura"

// 2 Coleta de dados

LEIA a opção do jogador 1

LEIA a opção do jogador 2

// 3 Tratamento de Dados de Entrada

CONVERTA a jogada do jogador 1 para letras minúsculas e sem espaços

CONVERTA a jogada do jogador 2 para letras minúsculas e sem espaços

ESCREVA "O jogador 1 escolheu":, jogada_jogador1

ESCREVA "O jogador 2 escolheu":, jogada_jogador2

// 4 Resultados do Jogo

SE jogada_jogador1 == jogada_jogador2

    ESCREVA: "Resultado: EMPATE!"

SENÃO SE (jogada_jogador1 == "papel" E jogada_jogador2 == "pedra") OU (jogada_jogador1 == "tesoura" E jogada_jogador2 == "papel") OU (jogada_jogador1 == "pedra" E jogada_jogador2 == "tesoura")

    ESCREVA: "Resultado: VITÓRIA DO JOGADOR 1!"

SENÃO

    ESCREVA: "Resultado: VITÓRIA DO JOGADOR 2!"

FIM




#### Pseudocódigo convertido para código Python

```python
# 1 Introdução

print("-------------------------------------------------")
print("--- Jogo Pedra, Papel e Tesoura (2 Jogadores) ---")
print("-------------------------------------------------")
print("Bem-vindos! Para começar, escolham entre uma das opções.")

# Uso de tupla pois as opções válidas são imutáveis.
opcoes_validas = ("pedra", "papel", "tesoura")
print(f"Opções válidas: {opcoes_validas}")
print("_" * 25) # Imprime uma linha para separar.

# 2 Coleta de Dados

jogada_jogador1_inicial = input("Qual a sua escolha, jogador 1?")
jogada_jogador2_inicial = input("Qual a sua escolha, jogador 2?")

# 3 Tratamento de Dados de Entrada 

jogada_jogador1 = jogada_jogador1_inicial.lower().strip() # lower() para deixar a frase em minúsculas e strip() para deixar a frase sem espaço, evitando erros de escrita.
jogada_jogador2 = jogada_jogador2_inicial.lower().strip()


print(f"O jogador 1 escolheu: {jogada_jogador1}")
print(f"O jogador 2 escolheu: {jogada_jogador2}")

print("_" * 25)


# 4 Lógica e resultados do jogo

# Primeiro checamos se a opção escolhida pelos jogadores são válidas
if jogada_jogador1 not in opcoes_validas or jogada_jogador2 not in opcoes_validas:
    print("Uma ou ambas as jogadas são inválidas! Por favor, use apenas 'pedra', 'papel' ou 'tesoura'")

elif jogada_jogador1 == jogada_jogador2:
    print("Resultado: Empate! Tentem novamente.")

elif (jogada_jogador1 == "pedra" and jogada_jogador2 == "tesoura") or \
     (jogada_jogador1 == "tesoura" and jogada_jogador2 == "papel")  or \
     (jogada_jogador1 == "papel" and jogada_jogador2 == "pedra"):
    print("Vitória do jogador 1! Parabéns!")

else:
    print("Vitória do jogador 2! Parabéns!")

print("--- Fim do jogo ---")





```
